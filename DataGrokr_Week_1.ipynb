```python
import csv

GRADE_SCALE = {
    "A+": 90,
    "A": 80,
    "B": 70,
    "C": 60,
    "D": 50,
    "F": 0
}

def calculate_grade(mark):
    if mark >= GRADE_SCALE["A+"]:
        return "A+"
    elif mark >= GRADE_SCALE["A"]:
        return "A"
    elif mark >= GRADE_SCALE["B"]:
        return "B"
    elif mark >= GRADE_SCALE["C"]:
        return "C"
    elif mark >= GRADE_SCALE["D"]:
        return "D"
    else:
        return "F"

def calculate_average(marks):
    if len(marks) == 0:
        return 0
    return sum(marks) / len(marks)

def get_performance(average):
    if average >= 90:
        return "Excellent"
    elif average >= 80:
        return "Very Good"
    elif average >= 70:
        return "Good"
    elif average >= 50:
        return "Pass"
    else:
        return "Needs Improvement"

def get_valid_mark(subject):
    while True:
        try:
            mark = float(input(f"Enter marks for {subject} (0-100): "))
            if mark < 0 or mark > 100:
                print("Marks must be between 0 and 100.")
                continue
            return mark
        except ValueError:
            print("Invalid input. Please enter a number.")

def create_student():
    print("Enter Student Details")

    name = input("Enter student name: ").strip()
    student_id = input("Enter student ID: ").strip()

    subjects = (
        "Python",
        "Data Structures",
        "Database",
        "Computer Networks",
        "Communication Skills"
    )

    marks = {}

    for subject in subjects:
        marks[subject] = get_valid_mark(subject)

    mark_list = list(marks.values())
    total = sum(mark_list)
    average = calculate_average(mark_list)

    grades = set()

    for mark in mark_list:
        grades.add(calculate_grade(mark))

    result = "PASS"

    for mark in mark_list:
        if mark < 50:
            result = "FAIL"
            break

    return {
        "name": name,
        "id": student_id,
        "marks": marks,
        "total": total,
        "average": average,
        "grades": grades,
        "result": result
    }

def display_result(student):
    print("\nGRADE REPORT")
    print(f"Student Name: {student['name']}")
    print(f"Student ID: {student['id']}")

    for subject, mark in student["marks"].items():
        grade = calculate_grade(mark)
        print(f"{subject}: {mark:.2f} - {grade}")

    print(f"Total Marks: {student['total']:.2f}")
    print(f"Average: {student['average']:.2f}")
    print(f"Performance: {get_performance(student['average'])}")
    print(f"Result: {student['result']}")

def save_to_text(student, filename="grades.txt"):
    try:
        with open(filename, "a", encoding="utf-8") as file:
            file.write("GRADE REPORT\n")
            file.write(f"Student Name: {student['name']}\n")
            file.write(f"Student ID: {student['id']}\n")

            for subject, mark in student["marks"].items():
                grade = calculate_grade(mark)
                file.write(f"{subject}: {mark:.2f} - {grade}\n")

            file.write(f"Total Marks: {student['total']:.2f}\n")
            file.write(f"Average: {student['average']:.2f}\n")
            file.write(
                f"Performance: {get_performance(student['average'])}\n"
            )
            file.write(f"Result: {student['result']}\n")
            file.write("\n")

        print(f"Result saved to {filename}")

    except IOError as error:
        print(f"Error saving file: {error}")

    finally:
        print("Text file operation completed.")

def save_to_csv(student, filename="grades.csv"):
    try:
        try:
            with open(filename, "r", encoding="utf-8") as file:
                file_exists = file.read(1) != ""
        except FileNotFoundError:
            file_exists = False

        with open(filename, "a", newline="", encoding="utf-8") as file:
            writer = csv.writer(file)

            if not file_exists:
                writer.writerow([
                    "Student ID",
                    "Student Name",
                    "Python",
                    "Data Structures",
                    "Database",
                    "Computer Networks",
                    "Communication Skills",
                    "Total",
                    "Average",
                    "Result"
                ])

            writer.writerow([
                student["id"],
                student["name"],
                student["marks"]["Python"],
                student["marks"]["Data Structures"],
                student["marks"]["Database"],
                student["marks"]["Computer Networks"],
                student["marks"]["Communication Skills"],
                student["total"],
                student["average"],
                student["result"]
            ])

        print(f"Result saved to {filename}")

    except IOError as error:
        print(f"Error saving CSV file: {error}")

    finally:
        print("CSV file operation completed.")

def view_saved_results(filename="grades.txt"):
    try:
        with open(filename, "r", encoding="utf-8") as file:
            content = file.read()

            if content.strip():
                print(content)
            else:
                print("No saved results found.")

    except FileNotFoundError:
        print("No result file exists yet.")

    except IOError as error:
        print(f"Error reading file: {error}")

    finally:
        print("File reading operation completed.")

def display_grade_scale():
    print("GRADE SCALE")

    for grade, minimum_mark in GRADE_SCALE.items():
        print(f"{grade}: {minimum_mark}+")

def main():
    students = []

    menu_options = ("1", "2", "3", "4", "5")

    while True:
        print("\nCLI GRADE CALCULATOR")
        print("1. Calculate Student Grade")
        print("2. Display Grade Scale")
        print("3. View Saved Results")
        print("4. View Students in Current Session")
        print("5. Exit")

        choice = input("Enter your choice: ").strip()

        if choice not in menu_options:
            print("Invalid choice. Please select 1-5.")
            continue

        if choice == "1":
            student = create_student()
            students.append(student)
            display_result(student)

            save_choice = input(
                "Do you want to save this result? (yes/no): "
            ).strip().lower()

            if save_choice == "yes":
                save_to_text(student)
                save_to_csv(student)

        elif choice == "2":
            display_grade_scale()

        elif choice == "3":
            view_saved_results()

        elif choice == "4":
            if len(students) == 0:
                print("No students calculated in this session.")
            else:
                for index, student in enumerate(students, start=1):
                    print(
                        f"{index}. {student['name']} | "
                        f"{student['id']} | "
                        f"Average: {student['average']:.2f} | "
                        f"{student['result']}"
                    )

        elif choice == "5":
            print("Thank you for using CLI Grade Calculator!")
            break

if __name__ == "__main__":
    try:
        main()
    except KeyboardInterrupt:
        print("\nProgram interrupted by user.")
    except Exception as error:
        print(f"Unexpected error: {error}")
    finally:
        print("Program closed.")
```
